Question 1: A die is rolled. What is the probability of getting:

(a) An even number

(b) A number greater than 4


ans  

(a) Probability of Getting an Even Number
Answer: 1/2 or 0.5 or 50%
Even numbers on die: 2, 4, 6 (3 outcomes out of 6)
P(Even) = 3/6 = 1/2

(b) Probability of Getting a Number Greater Than 4
Answer: 1/3 or 0.333 or 33.3%
Numbers > 4: 5, 6 (2 outcomes out of 6)
P(>4) = 2/6 = 1/3

Question 3: A bag has 3 red and 2 blue balls. If one ball is drawn randomly and is red, what is the probability
that the next ball is also red (without replacement)?

ans

Initial situation:

3 red balls
2 blue balls
Total: 5 balls

After drawing 1 red ball:

Red balls remaining: 3 - 1 = 2 red
Blue balls remaining: 2 (unchanged)
Total balls remaining: 4

Probability of drawing red next:
P(2nd red | 1st red) = Remaining red / Remaining total
                     = 2/4
                     = 1/2
                     = 0.5
                     = 50%

✅ Guarantees equal representation

Question 5: The average height of 1000 students = 160 cm. A sample of 100 students shows an average height
= 158 cm.

Find the sampling error.

ANS

Sampling Error = |Sample Mean - Population Mean|
               = |158 - 160|
               = |-2|
               = 2 cm

Question 6: The population mean salary is ₹50,000 with σ = ₹5,000. If we take a sample of 100 employees, what
is the standard error of the mean (SEM)   

ANS


SEM = 5,000 / √100
    = 5,000 / 10
    = ₹500

Question 7: In a group of 100 students:

40 like Cricket (C)

30 like Football (F)

10 like both Cricket and Football

Find the probability that a student likes at least one sport.

ANS

Students who like at least one sport = 30 + 40 - 10 = 60

P(at least one sport) = 60/100 = 3/5 = 0.6 = 60%

Question 8: From a deck of 52 cards, two cards are drawn without replacement. What is the probability that
both are Aces?

ANS

P(1st Ace) = 4/52

P(2nd Ace | 1st Ace) = 3/51
(After drawing 1 Ace, 3 Aces remain out of 51 cards)

P(Both Aces) = (4/52) × (3/51)
             = 12/2652
             = 1/221
             = 0.0045
             = 0.45%

             

Question 9: A factory produces bulbs with 2% defective rate. If 5 bulbs are chosen at random, what is the
probability that all are non-defective?

ANS

P(All 5 non-defective) = P(non-defective)^5
                       = (0.98)^5


                       (0.98)^5 = 0.98 × 0.98 × 0.98 × 0.98 × 0.98
         = 0.9039
         = 90.39%   

Question 10: Differentiate between discrete and continuous random variables with examples.  

ANS

DISCRETE RANDOM VARIABLES


✅ Countable values
✅ Finite or countably infinite outcomes
✅ Can be listed: {1, 2, 3, ...}
✅ Probability at specific points
✅ No values between consecutive outcomes


CONTINUOUS RANDOM VARIABLES


✅ Infinite uncountable values
✅ Any value in a range
✅ Cannot be listed
✅ Probability in intervals
✅ Probability at specific point = 0
✅ Represented by curves/intervals
